```python
import json
import os

def find_image_file(txt_file_path, res_directory):
    file_name_without_extension = os.path.splitext(os.path.basename(txt_file_path))[0]
    image_extensions = ['.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp', '.svg']
    
    for extension in image_extensions:
        image_file_path = os.path.join(res_directory, file_name_without_extension + extension)
        if os.path.exists(image_file_path):
            return image_file_path
    return None

def replace_image_tag_with_filename(txt_content, image_file_path):
    json_structure = {
        "conversations": [
            {
                "from": "human",
                "value": ""
            },
            {
                "from": "gpt",
                "value": txt_content
            }
        ],
        "images": ""
    }

    if image_file_path:
        json_structure['images'] = image_file_path
        json_structure['conversations'][0]['value'] = f"Summary the topology image below: <image>"
    else:
        json_structure['images'] = "<image> Image file not found"
        json_structure['conversations'][0]['value'] = "Summary the topology image below: <image> Image file not found"

    return json_structure

directory = "/home/yantao/llm2024/topo2text/data/res_summary_len8"
pic_directory = "/home/yantao/llm2024/topo2text/data/images/normal"
save_directory = "/home/yantao/llm2024/topo2text/LLaMA-Factory/data/"

output_json_path = os.path.join(save_directory, 'mllm_topo2text_demo.json')

with open(output_json_path, 'w', encoding='utf-8') as json_file:
    json_file.write('[')

    first_entry = True
    for filename in os.listdir(directory):
        if filename.endswith('.txt'):
            txt_file_path = os.path.join(directory, filename)
            
            with open(txt_file_path, 'r', encoding='utf-8') as file:
                txt_content = file.read()
            
            image_file_path = find_image_file(txt_file_path, pic_directory)

            output_json_data = replace_image_tag_with_filename(txt_content, image_file_path)

            if not first_entry:
                json_file.write(',\n')
            first_entry = False
            
            json.dump(output_json_data, json_file, ensure_ascii=False, indent=2)
    
    json_file.write('\n]')

print(f"Saved data to {output_json_path}")
```
